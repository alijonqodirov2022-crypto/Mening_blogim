# **1- MASALA:** **Vacuum - Cleaner World agentini yaratish. (2 ta xona uchun)**



---
##**Maqsad:**
##2 ta xonali (A va B) muhitda tozalikni saqlash.

## **Vazifa:**

## Agent hozir qaysi xonada ekanligini va xonaning holatini (toza yoki kir) aniqlashi kerak.

* ## *Agar xona "Kir" bo'lsa -> "Tozalamoq".*

* ## *Agar xona "Toza" bo'lsa va A da bo'lsa -> "O'ngga yurish".*

* ## *Agar xona "Toza" bo'lsa va B da bo'lsa -> "Chapga yurish".*


#**⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>xonalar = {"A": "iflos", "B": "toza"}</code>

* ### <code>agent_joylashuvi = "A"</code>

##***Klasslar:***

* ### <code>Muhit</code>
  * ### <code>perseptsiya_ol()</code>
  * ### <code>harakat_qollash(harakat)</code>
* ### <code>RefleksAgent</code>
  * ### <code>qaror_qabul_qil(perseptsiya)</code>

##***Simulyatsiya:***
* ### 20 tadan kam emas
 ## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.

```python
class Muhit:
    def __init__(self):

        self.xonalar = {"A": "iflos", "B": "toza"}
        self.agent_joylashuvi = "A"

    def perseptsiya_ol(self):

        return (self.agent_joylashuvi, self.xonalar[self.agent_joylashuvi])

    def harakat_qollash(self, harakat):

        if harakat == "TOZALAMOQ":
            self.xonalar[self.agent_joylashuvi] = "toza"
            print(f"  -> {self.agent_joylashuvi} xonasi tozalandi!")
        elif harakat == "O'NGGA_YURISH":
            self.agent_joylashuvi = "B"
            print(f"  -> Agent B xonaga o'tdi.")
        elif harakat == "CHAPGA_YURISH":
            self.agent_joylashuvi = "A"
            print(f"  -> Agent A xonaga qaytdi.")


class RefleksAgent:
    def qaror_qabul_qil(self, perseptsiya):
        # Percept asosida qoida tanlaydi
        joylashuv, holat = perseptsiya
        if holat == "iflos":
            return "TOZALAMOQ"
        elif joylashuv == "A":
            return "O'NGGA_YURISH"
        elif joylashuv == "B":
            return "CHAPGA_YURISH"

muhit = Muhit()
agent = RefleksAgent()

print("=" * 55)
print("MASALA 1: Vacuum Cleaner Agent - 10 qadam simulyatsiya")
print("=" * 55)

for qadam in range(1, 21):
    perseptsiya = muhit.perseptsiya_ol()
    harakat = agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: Joylashuv={perseptsiya[0]}, Holat={perseptsiya[1]:5s} --> Harakat: {harakat}")
    muhit.harakat_qollash(harakat)

```

# **2-MASALA: Haroratni nazorat qiluvchi agentni yaratish.**


---


## **Maqsad:**

##Sonli ma’lumot (harorat) asosida refleks qaror chiqarishni o‘rganish.

## **Vazifa:**

## Xona haroratiga qarab isitish yoki sovitish qurilmasini boshqarish.

* ### harorat < 18 → **ISITISH_YOQ**

* ### harorat > 26 → **SOVITISH_YOQ**

* ###  Aks holda → **KUTISH**

## **⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>harorat</code> (10 - 35 oralig'ida)



##***Klasslar:***

* ### <code>HaroratMuhiti</code>

* ### <code>HaroratAgenti</code>

##***Simulyatsiya:***
* ### 20 tadan kam emas

 ## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.

```python
import random

class HaroratMuhiti:
    def __init__(self):

        self.harorat = random.randint(10, 35)

    def perseptsiya_ol(self):
        return self.harorat

    def harakat_qollash(self, harakat):
        print(f"  -> Qurilma buyrug'i: {harakat}")


class HaroratAgenti:
    def qaror_qabul_qil(self, perseptsiya):
        harorat = perseptsiya
        if harorat < 18:
            return "ISITISH_YOQ"
        elif harorat > 26:
            return "SOVITISH_YOQ"
        else:
            return "KUTISH"

h_muhit = HaroratMuhiti()
h_agent = HaroratAgenti()

print("=" * 55)
print("MASALA 2: Harorat Agenti - 20 qadam simulyatsiya")
print("=" * 55)

for qadam in range(1, 21):
    h_muhit.harorat = random.randint(10, 35)
    perseptsiya = h_muhit.perseptsiya_ol()
    harakat = h_agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: Harorat={perseptsiya}°C --> Harakat: {harakat}")
    h_muhit.harakat_qollash(harakat)
```

# **3-MASALA: Aqlli sug‘orish agentini yaratish**

---


## **Maqsad:**

## Agentga mantiqiy va (and) operatori hamda inkor (not) tushunchasini qo‘llash.

## **Vazifa:**

## Tuproq namligi va yomg‘ir bor-yo‘qligiga qarab suv nasosini boshqarish.

* ### yomgir = True → **NASOSNI_OCHIR**

* ### namlik < 30 va yomgir = False → **NASOSNI_YOQ**

* ###  namlik >= 30 va yomgir == False → NASOS_OCH → **NASOSNI_OCHIR**
* ### Aks holda → **TEKSHIRISH_REJIMI**

## **⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>namlik</code> (0-100 oralig'idagi son)
* ### <code>yomgir</code> (True yoki False)



##***Klasslar:***

* ### <code>BogMuhiti</code>

* ### <code>SugorishAgenti</code>

##***Simulyatsiya:***
* ### 20 tadan kam emas


 ## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.


```python
class BogMuhiti:
    def __init__(self):
        self.namlik = random.randint(0, 100)
        self.yomgir = random.choice([True, False])

    def perseptsiya_ol(self):
        return (self.namlik, self.yomgir)

    def harakat_qollash(self, harakat):
        print(f"  -> Nasos holati: {harakat}")


class SugorishAgenti:
    def qaror_qabul_qil(self, perseptsiya):
        namlik, yomgir = perseptsiya
        if yomgir:
            return "NASOSNI_OCHIR"
        elif namlik < 30 and not yomgir:
            return "NASOSNI_YOQ"
        elif namlik >= 30 and not yomgir:
            return "NASOSNI_OCHIR"
        else:
            return "TEKSHIRISH_REJIMI"

b_muhit = BogMuhiti()
b_agent = SugorishAgenti()

print("=" * 60)
print("MASALA 3: Sug'orish Agenti - 20 qadam simulyatsiya")
print("=" * 60)

for qadam in range(1, 21):
    b_muhit.namlik = random.randint(0, 100)
    b_muhit.yomgir = random.choice([True, False])
    perseptsiya = b_muhit.perseptsiya_ol()
    harakat = b_agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: Namlik={perseptsiya[0]:3d}%, Yomg'ir={str(perseptsiya[1]):5s} --> Harakat: {harakat}")
    b_muhit.harakat_qollash(harakat)
```

# **4-MASALA: Xavfsizlik qorovuli agentini yaratish.**


---


## **Maqsad:**

## Agent ikki xil sensordan kelgan ma'lumotlarni birlashtirib qaror qabul qilish.

## **Vazifa:**

## Harakat sensori va eshik qulfi holatiga qarab signal chalish.

* ### harakat bor va qulf ochiq bo'lsa  → **SIGNAL_CHALISH**

* ### harakat bor va qulf yopiq bo'lsa → **SALOMLASHISH**

* ###  harakat yo'q → **JIM_TURISH**
* ### Aks holda → **KUZATISH**

## **⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>harakat</code> (True yoki False)
* ### <code>qulf_ochiq</code> (True yoki False)



##***Klasslar:***

* ### <code>UyMuhiti</code>

* ### <code>QorovulAgenti</code>


##***Simulyatsiya:***
* ### 20 tadan kam emas

 ## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.



```python
class UyMuhiti:
    def __init__(self):
        self.harakat = random.choice([True, False])
        self.qulf_ochiq = random.choice([True, False])

    def perseptsiya_ol(self):
        return (self.harakat, self.qulf_ochiq)

    def harakat_qollash(self, harakat):
        print(f"  -> Qorovul harakati: {harakat}")


class QorovulAgenti:
    def qaror_qabul_qil(self, perseptsiya):
        harakat, qulf_ochiq = perseptsiya
        if harakat and qulf_ochiq:
            return "SIGNAL_CHALISH"
        elif harakat and not qulf_ochiq:
            return "SALOMLASHISH"
        elif not harakat:
            return "JIM_TURISH"
        else:
            return "KUZATISH"

u_muhit = UyMuhiti()
u_agent = QorovulAgenti()

print("=" * 65)
print("MASALA 4: Xavfsizlik Qorovul Agenti - 20 qadam simulyatsiya")
print("=" * 65)

for qadam in range(1, 21):
    u_muhit.harakat = random.choice([True, False])
    u_muhit.qulf_ochiq = random.choice([True, False])
    perseptsiya = u_muhit.perseptsiya_ol()
    harakat = u_agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: Harakat={str(perseptsiya[0]):5s}, Qulf_ochiq={str(perseptsiya[1]):5s} --> {harakat}")
    u_muhit.harakat_qollash(harakat)

```

# **5-MASALA: Liftni boshqaruvchi agentni yaratish.**

---


## **Maqsad:**

## Agent joriy holat va maqsadli holatni solishtiradi.

## **Vazifa:**

## Lift turgan qavat va chaqirilgan qavatga qarab harakat yo‘nalishini aniqlash.

* ### **YUQORIGA**
* ### **PASTGA**
* ### **ESHIKNI_OCH**
* ### Aks holda → **KUTISH**


## **⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>lift_qavati</code> (1-9 oralig'ida)
* ### <code>chaqiruv_qavati</code> (1-9 oralig'ida)



##***Klasslar:***

* ### <code>BinoMuhiti</code>

* ### <code>LiftAgenti</code>

##***Simulyatsiya:***
* ### 20 tadan kam emas


 ## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.

```python
class BinoMuhiti:
    def __init__(self):
        self.lift_qavati = random.randint(1, 9)
        self.chaqiruv_qavati = random.randint(1, 9)

    def perseptsiya_ol(self):
        return (self.lift_qavati, self.chaqiruv_qavati)

    def harakat_qollash(self, harakat):
        print(f"  -> Lift harakati: {harakat}")


class LiftAgenti:
    def qaror_qabul_qil(self, perseptsiya):
        lift_qavati, chaqiruv_qavati = perseptsiya
        if lift_qavati < chaqiruv_qavati:
            return "YUQORIGA"
        elif lift_qavati > chaqiruv_qavati:
            return "PASTGA"
        elif lift_qavati == chaqiruv_qavati:
            return "ESHIKNI_OCH"
        else:
            return "KUTISH"


l_muhit = BinoMuhiti()
l_agent = LiftAgenti()

print("=" * 60)
print("MASALA 5: Lift Agenti - 20 qadam simulyatsiya")
print("=" * 60)

for qadam in range(1, 21):
    l_muhit.lift_qavati = random.randint(1, 9)
    l_muhit.chaqiruv_qavati = random.randint(1, 9)
    perseptsiya = l_muhit.perseptsiya_ol()
    harakat = l_agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: Lift={perseptsiya[0]}-qavat, Chaqiruv={perseptsiya[1]}-qavat --> {harakat}")
    l_muhit.harakat_qollash(harakat)
```

# **6-MASALA: Server yuklamasini balanslovchi agent yaratish.**

---

## **Maqsad:**

## Agent resurslarni boshqarish uchun chegaraviy qiymatlar (thresholds) bilan ishlaydi.

## **Vazifa:**

## Protsessor (CPU) yuklamasiga qarab qo‘shimcha serverlarni yoqish yoki o‘chirish.

* ### **SERVER_QOSHISH**
* ### **SERVER_OCHIRISH**
* ### **STABIL_HOLAT**


## **⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>cpu_yuklama</code> (0-100 foiz oralig'ida)

##***Klasslar:***

* ### <code>DataMarkazMuhiti</code>

* ### <code>BalansAgenti</code>

##***Simulyatsiya:***
* ### 20 tadan kam emas

 ## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.

```python
class DataMarkazMuhiti:
    def __init__(self):
        self.cpu_yuklama = random.randint(0, 100)

    def perseptsiya_ol(self):
        return self.cpu_yuklama

    def harakat_qollash(self, harakat):
        print(f"  -> Server holati: {harakat}")


class BalansAgenti:
    def qaror_qabul_qil(self, perseptsiya):
        cpu_yuklama = perseptsiya
        if cpu_yuklama > 80:
            return "SERVER_QOSHISH"
        elif cpu_yuklama < 20:
            return "SERVER_OCHIRISH"
        else:
            return "STABIL_HOLAT"


d_muhit = DataMarkazMuhiti()
d_agent = BalansAgenti()

print("=" * 60)
print("MASALA 6: Server Balans Agenti - 20 qadam simulyatsiya")
print("=" * 60)

for qadam in range(1, 21):
    d_muhit.cpu_yuklama = random.randint(0, 100)
    perseptsiya = d_muhit.perseptsiya_ol()
    harakat = d_agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: CPU={perseptsiya:3d}% --> Harakat: {harakat}")
    d_muhit.harakat_qollash(harakat)

```

# **7-MASALA: Batareya tejash tizimini yaratish**

---

## **Maqsad:**

## Tizim telefon quvvati va quvvatlash holatini inobatga olib rejimni o‘zgartirish.

## **Vazifa:**

## Telefon quvvati kam qolsa, tejash rejimini yoqish, lekin quvvatlanayotgan bo‘lsa yoqmaslik.

* ### **TEJASH_REJIMI**
* ### **NORMAL_REJIM**
* ### **NORMAL_REJIM**
* ### Aks holda → **XATOLIK**

## **⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>quvvat</code> (0-100 foiz oralig'ida)
* ### <code>quvvatlanish</code> (True yoki False)

##***Klasslar:***

* ### <code>TelefonMuhiti</code>

* ### <code>TelefonMuhiti</code>

##***Simulyatsiya:***
* ### 20 tadan kam emas

 ## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.




```python
class TelefonMuhiti:
    def __init__(self):
        self.quvvat = random.randint(0, 100)
        self.quvvatlanish = random.choice([True, False])

    def perseptsiya_ol(self):
        return (self.quvvat, self.quvvatlanish)

    def harakat_qollash(self, harakat):
        print(f"  -> Telefon rejimi: {harakat}")

    def qaror_qabul_qil(self, perseptsiya):

        quvvat, quvvatlanish = perseptsiya
        if quvvat < 20 and not quvvatlanish:
            return "TEJASH_REJIMI"
        elif quvvat < 20 and quvvatlanish:
            return "NORMAL_REJIM"
        elif quvvat >= 20:
            return "NORMAL_REJIM"
        else:
            return "XATOLIK"


t_muhit = TelefonMuhiti()

print("=" * 65)
print("MASALA 7: Batareya Tejash Tizimi - 20 qadam simulyatsiya")
print("=" * 65)

for qadam in range(1, 21):
    t_muhit.quvvat = random.randint(0, 100)
    t_muhit.quvvatlanish = random.choice([True, False])
    perseptsiya = t_muhit.perseptsiya_ol()
    harakat = t_muhit.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: Quvvat={perseptsiya[0]:3d}%, Quvvatlanish={str(perseptsiya[1]):5s} --> {harakat}")
    t_muhit.harakat_qollash(harakat)
```

```python
class PochtaMuhiti:
    def __init__(self, xabar):
        self.xabar = xabar

    def perseptsiya_ol(self):
        return self.xabar

    def harakat_qollash(self, harakat):
        print(f"  -> Xabar joylashtirildi: [{harakat}]")


class SpamAgenti:
    def qaror_qabul_qil(self, perseptsiya):
        xabar = perseptsiya.lower()
        if "yutuq" in xabar or "tekin" in xabar:
            return "SPAM_PAPKA"
        elif "hisobot" in xabar or "ish" in xabar:
            return "MUHIM_PAPKA"
        else:
            return "KIRUVCHI_PAPKA"


xabarlar = [
    "Siz katta yutuq qozondiniz, sovg'angizni oling!",
    "Bugungi ish hisoboti tayyor va yuborildi.",
    "Salom, qanday yuribsiz? Yaxshi kunlar!",
    "Tekin kurs! Ro'yxatdan o'ting hoziroq!",
    "Loyiha ish rejasi haqida muhokama kerak.",
    "Kechqurun uchrashamizmi, do'stim?",
    "Haftalik hisobot uchun ma'lumotlar kerak.",
    "Yutuq! Million dollar yutdingiz, shoshiling!",
    "Ertaga ish boshlanadi, tayyor bo'ling.",
    "Yangi rasm yuklandim, qarang!",
    "Tekin internet topildi, bu linkni bos!",
    "Xavfsizlik hisoboti tayyorlandi.",
    "Dam olish kunlari qanday o'tdi?",
    "Yutuqli aksiya: bepul mahsulot oling!",
    "Ish joyida yangi qoidalar joriy qilindi.",
    "Tug'ilgan kuningiz bilan tabriklayman!",
    "Tekin taklif: bugun faqat 100 ta!",
    "Moliyaviy hisobot yuborildi, tekshiring.",
    "Kutubxonaga kitob qaytarishni unutmang.",
    "Yutuq o'yinida qatnashing va g'olib bo'ling!"
]

s_agent = SpamAgenti()

print("=" * 65)
print("MASALA 8: Spam Aniqlovchi Agenti - 20 ta xabar tekshiruvi")
print("=" * 65)

for qadam, xabar in enumerate(xabarlar, 1):
    p_muhit = PochtaMuhiti(xabar)
    perseptsiya = p_muhit.perseptsiya_ol()
    harakat = s_agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: '{xabar[:45]}...' --> {harakat}")
    p_muhit.harakat_qollash(harakat)

```

## **9-MASALA: Aqlli avtomobil parkovka agentini yaratish.**

---

## **Maqsad:**

## Masofa datchiklari (sensorlari) asosida obyektgacha bo‘lgan oraliqni aniqlash va xavfsiz to‘xtashni ta’minlash.

## **Vazifa:**

## Avtomobil orqaga harakatlanayotganda devorgacha bo‘lgan masofaga qarab tormozni boshqarish.

* ### masofa <= 0.5 → **TOXTASH** (Avariya xavfi)

* ### masofa > 0.5 va masofa <= 2.0 → **SEKINLASHISH** (Ogohlantirish signali)

* ### Aks holda → **ORQAGA_YURISH**

## **⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>masofa</code> (float - metr hisobida, masalan: 5.4, 1.2, 0.3)

##***Klasslar:***

* ### <code>GarajMuhiti</code>

* ### <code>ParkovkaAgenti</code>

##***Simulyatsiya:***
* ### 20 tadan kam emas

 ## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.


```python
class GarajMuhiti:
    def __init__(self):
        self.masofa = round(random.uniform(0.1, 6.0), 1)

    def perseptsiya_ol(self):
        return self.masofa

    def harakat_qollash(self, harakat):
        print(f"  -> Avtomobil harakati: {harakat}")


class ParkovkaAgenti:
    def qaror_qabul_qil(self, perseptsiya):
        masofa = perseptsiya
        if masofa <= 0.5:
            return "TOXTASH"
        elif masofa > 0.5 and masofa <= 2.0:
            return "SEKINLASHISH"
        else:
            return "ORQAGA_YURISH"


g_muhit = GarajMuhiti()
g_agent = ParkovkaAgenti()

print("=" * 60)
print("MASALA 9: Parkovka Agenti - 20 qadam simulyatsiya")
print("=" * 60)

for qadam in range(1, 21):
    g_muhit.masofa = round(random.uniform(0.1, 6.0), 1)
    perseptsiya = g_muhit.perseptsiya_ol()
    harakat = g_agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: Masofa={perseptsiya:4.1f}m --> Harakat: {harakat}")
    g_muhit.harakat_qollash(harakat)

```

## **10-MASALA: Aqlli muzlatkich agentini yaratish.**

---

## **Maqsad:**

## Bir nechta (3 ta) sensordan kelgan ma'lumotlarni ustuvorlik bo‘yicha tahlil qilish.

## **Vazifa:**

## Muzlatkich eshigi, ichki harorat va mahsulot miqdoriga qarab, mos xabarni chiqarish.

* ### **SIGNAL_CHALISH** (Eng muhimi)
* ### **SOVITISHNI_KUCHAYTIRISH**
* ### **NORMAL_REJIM**
* ### Aks holda → **BUYURTMA_BERISH**
* ### Aks holda → **NORMAL_HOLAT**

## **⚙️ Bajarish tartibi**

##***Ma'lumotlar:***

* ### <code>eshik_ochiq</code> (True yoki False)

* ### <code>harorat</code> (butun son, masalan: 4, 8, -2)

* ### <code>mahsulot_soni</code> (butun son)

##***Klasslar:***

* ### <code>OshxonaMuhiti</code>

* ### <code>MuzlatkichAgenti</code>

##***Simulyatsiya:***
* ### 20 tadan kam emas

## **🚫Eslatma:** Masaladagi o'zgaruvchilar, ma'lumotlar, klasslar bajarish tartibida keltirilgan nomlar bilan nomlansin.


```python
class OshxonaMuhiti:
    def __init__(self):
        self.eshik_ochiq = random.choice([True, False])
        self.harorat = random.randint(-5, 15)
        self.mahsulot_soni = random.randint(0, 20)

    def perseptsiya_ol(self):
        return (self.eshik_ochiq, self.harorat, self.mahsulot_soni)

    def harakat_qollash(self, harakat):
        print(f"  -> Muzlatkich holati: {harakat}")


class MuzlatkichAgenti:
    def qaror_qabul_qil(self, perseptsiya):
        eshik_ochiq, harorat, mahsulot_soni = perseptsiya

        if eshik_ochiq:
            return "SIGNAL_CHALISH"
        elif harorat > 5:
            return "SOVITISHNI_KUCHAYTIRISH"
        elif harorat <= 5 and mahsulot_soni > 0:
            return "NORMAL_REJIM"
        elif mahsulot_soni == 0:
            return "BUYURTMA_BERISH"
        else:
            return "NORMAL_HOLAT"


o_muhit = OshxonaMuhiti()
o_agent = MuzlatkichAgenti()

print("=" * 70)
print("MASALA 10: Muzlatkich Agenti - 20 qadam simulyatsiya")
print("=" * 70)

for qadam in range(1, 21):
    o_muhit.eshik_ochiq = random.choice([True, False])
    o_muhit.harorat = random.randint(-5, 15)
    o_muhit.mahsulot_soni = random.randint(0, 20)
    perseptsiya = o_muhit.perseptsiya_ol()
    harakat = o_agent.qaror_qabul_qil(perseptsiya)
    print(f"Qadam {qadam:2d}: Eshik={str(perseptsiya[0]):5s}, Harorat={perseptsiya[1]:3d}°C, Mahsulot={perseptsiya[2]:2d} --> {harakat}")
    o_muhit.harakat_qollash(harakat)

```
